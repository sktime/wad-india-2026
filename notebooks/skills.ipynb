## Skills

- A folder with a `SKILL.md`: instructions for one task
- Agent sees only the description, loads the rest when a task matches
- Can bundle extra files: references, scripts

## Why they help

- Project knowledge the model lacks: conventions, gotchas, checks
- Same steps every time, fewer missed ones
- Agent runs the checks before handing over

## Write your own

`.claude/skills/add-model/SKILL.md`

```markdown
---
name: add-model
description: Add a new model to this package. Use when someone asks to implement or wrap a model.
---

1. Copy `templates/model.py`, fill in the todos
2. Add at least 2 test parameter sets
3. Run `pytest tests/test_models.py` before handing over
```

- `description`: when to load it
- Body: the steps, in plain words

You can get some skills useful for you work here: https://www.skills.sh/

## Watch out

- More tokens: every description is always in context, loaded bodies add more
- Vague description: loads when not needed, or never
- Others' skills can include scripts: read them before installing

## Agents: limits

- Hallucinate: made-up functions, parameters, tags
- Passing tests, wrong logic
- Extra code and changes outside the task
- Use them, but review the code yourself before pushing
