# Summary & get involved!

- unified interface for objects/estimators
- modular, composable, plug & play
- simple specification
- **user first, no lock-in**

### [www.sktime.net](https://www.sktime.net)

* forecasting with `sktime`

    * exogenous data
    * uni- and multivariate
    * probabilistic forecasting
    * hierarchical forecasting
    * feature engineering, pipelines
    * tuning - backtesting, autoML

* deep learning and foundation models

    * DL models, `pytorch-forecasting`
    * pre-training, global forecasting
    * foundation models

* `sktime` agentic features
    
    * forecaster harness - `AutoResearch`
    * skills
        * user skills
        * extender skills, extension template, API checking harness
        * developer skills
    * MCP
